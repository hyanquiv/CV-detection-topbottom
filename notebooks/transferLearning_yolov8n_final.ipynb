## Modificando Labels

```python
import os
import shutil
import yaml

# Diccionario de mapeo por carpeta
LABEL_MAPS = {
    "Clothes-1": ["Bottom", "Dress", "Top"],
    "Clothes-11": ["Jacket", "Jeans", "Jogger", "Polo", "Shirt", "Short", "T-Shirt", "Trouser"],
    "Clothes-5": ["Jacket", "Jeans", "Jogger", "Polo", "Shirt", "Short", "T-Shirt", "Trouser"],
    "Clothing-Detection-6": ["Exposed", "Female", "Flats", "High Heel", "Hoodie", "Jacket", "Long Pants",
                             "Long Shirt", "Long Skirt", "Male", "Shirt", "Short", "Short Skirt",
                             "SleevelessShirt", "Slipper", "Sport Shoes"],
    "clothes-1": ["Blazer", "Blouse", "Hoodie", "Jacket", "JumpSuit", "Longsleeve", "Other", "Pants",
                  "Shirt", "Shoes", "Shorts", "Skirt", "T-Shirt", "Top", "Underwear", "Vest", "cap", "onepiece"],
    "clothes-classfy-1": ["Tshirt", "dress", "jacket", "pants", "shirt", "short", "skirt", "sweater"],
    "clothes-detection-2": ["dress", "pants", "shirt", "short", "skirt", "sweater", "tshirt"]
}

TOP_LABELS = {
    "Top", "Polo", "Shirt", "T-Shirt", "Long Shirt", "SleevelessShirt", "Blouse",
    "Longsleeve", "Tshirt", "shirt", "tshirt", "Vest", "jacket", "Jacket", "Hoodie", "Blazer", "sweater"
}
BOTTOM_LABELS = {
    "Bottom", "Jeans", "Jogger", "Trouser", "Pants", "Short", "Shorts", "Skirt", "Long Pants",
    "Short Skirt", "Long Skirt", "pants", "short", "skirt"
}
IGNORED_LABELS = {"Dress", "dress", "onepiece", "JumpSuit", "Other", "Underwear"}

def build_class_map(label_list):
    mapping = {}
    for idx, label in enumerate(label_list):
        if label in TOP_LABELS:
            mapping[idx] = 1
        elif label in BOTTOM_LABELS:
            mapping[idx] = 2
        else:
            mapping[idx] = -1
    return mapping

def process_folder(folder):
    print(f"Procesando: {folder}")
    class_list = LABEL_MAPS[folder]
    mapping = build_class_map(class_list)
    for split in ['train', 'valid', 'test']:
        label_dir = os.path.join(folder, split, "labels")
        image_dir = os.path.join(folder, split, "images")
        if not os.path.exists(label_dir):
            continue

        for file in os.listdir(label_dir):
            if not file.endswith(".txt"):
                continue
            file_path = os.path.join(label_dir, file)
            new_lines = []
            with open(file_path, 'r') as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) < 5:
                        continue
                    cls_idx = int(parts[0])
                    new_cls = mapping.get(cls_idx, -1)
                    if new_cls != -1:
                        new_lines.append(f"{new_cls} {' '.join(parts[1:])}")

            if new_lines:
                with open(file_path, 'w') as f:
                    f.write("\n".join(new_lines) + "\n")
            else:
                # Eliminar label y su imagen correspondiente
                os.remove(file_path)
                img_file = os.path.splitext(file)[0]
                for ext in ['.jpg', '.jpeg', '.png']:
                    img_path = os.path.join(image_dir, img_file + ext)
                    if os.path.exists(img_path):
                        os.remove(img_path)
                        break

    # Actualizar data.yaml
    yaml_path = os.path.join(folder, "data.yaml")
    if os.path.exists(yaml_path):
        with open(yaml_path, 'w') as f:
            yaml.dump({'nc': 2, 'names': ['top', 'bottom']}, f)
        print(f"✔️  data.yaml actualizado en {folder}")

if __name__ == "__main__":
    for folder in LABEL_MAPS:
        if os.path.exists(folder):
            process_folder(folder)
        else:
            print(f"⚠️  Carpeta no encontrada: {folder}")
```

```python
import os
import random
from glob import glob
from collections import defaultdict

DATASET_DIR = "people-1"
TARGET_PERSON_TOTAL = 10000
SUBSETS = ["train", "valid", "test"]

def count_detections_by_subset():
    counts = defaultdict(int)
    files = defaultdict(list)

    for subset in SUBSETS:
        label_dir = os.path.join(DATASET_DIR, subset, "labels")
        for lbl_file in glob(os.path.join(label_dir, "*.txt")):
            with open(lbl_file, "r") as f:
                lines = [line.strip() for line in f if line.strip()]
                count = sum(1 for line in lines if line.startswith("0"))
                if count > 0:
                    files[subset].append((lbl_file, count))
                    counts[subset] += count
    return counts, files

def delete_files(files_to_delete):
    for label_file in files_to_delete:
        image_file = None
        base = os.path.splitext(label_file)[0]
        for ext in ['.jpg', '.jpeg', '.png']:
            candidate = base.replace("labels", "images") + ext
            if os.path.exists(candidate):
                image_file = candidate
                break
        if os.path.exists(label_file):
            os.remove(label_file)
        if image_file and os.path.exists(image_file):
            os.remove(image_file)

def main():
    print("🔍 Contando detecciones actuales por subset...")
    counts, files = count_detections_by_subset()
    total_person = sum(counts.values())

    print(f"\n📊 Detecciones actuales:")
    for subset in SUBSETS:
        print(f"  {subset}: {counts[subset]} detecciones")

    if total_person <= TARGET_PERSON_TOTAL:
        print("\n✅ Ya hay menos de 10k detecciones. No se elimina nada.")
        return

    print(f"\n⚠️ Se exceden las 15k detecciones ({total_person}), eliminando proporcionalmente...")

    # Calculamos cuántas detecciones debe tener cada subset tras el recorte
    target_per_subset = {
        subset: int((counts[subset] / total_person) * TARGET_PERSON_TOTAL)
        for subset in SUBSETS
    }

    for subset in SUBSETS:
        to_remove = counts[subset] - target_per_subset[subset]
        if to_remove <= 0:
            continue

        # Ordenar aleatoriamente archivos que contienen detecciones
        random.shuffle(files[subset])

        removed = 0
        files_to_delete = []
        for lbl_file, det_count in files[subset]:
            files_to_delete.append(lbl_file)
            removed += det_count
            if removed >= to_remove:
                break

        delete_files(files_to_delete)
        print(f"  🗑️  {subset}: Eliminadas {len(files_to_delete)} imágenes ({removed} detecciones)")

    print("\n✅ Reducción completada. Dataset balanceado con ~10k detecciones.")

if __name__ == "__main__":
    main()
```

## Unificando

```python
import os
import shutil
from pathlib import Path
import yaml

# Carpeta destino
MERGED_DIR = "merged-dataset"
SUBSETS = ["train", "valid", "test"]

# Carpeta adicional que contiene la clase "person"
PEOPLE_DIR = "people-1"

# Clases finales
FINAL_CLASSES = ["person", "top", "bottom"]

# Límite de longitud de nombre de archivo (nombre + extensión)
MAX_FILENAME_LENGTH = 255

def ensure_dirs():
    for subset in SUBSETS:
        os.makedirs(os.path.join(MERGED_DIR, subset, "images"), exist_ok=True)
        os.makedirs(os.path.join(MERGED_DIR, subset, "labels"), exist_ok=True)

def is_name_too_long(path):
    return len(os.path.basename(path)) > MAX_FILENAME_LENGTH

def copy_and_suffix_files(dataset_folder):
    for subset in SUBSETS:
        label_dir = os.path.join(dataset_folder, subset, "labels")
        image_dir = os.path.join(dataset_folder, subset, "images")
        if not os.path.exists(label_dir):
            continue

        for label_file in os.listdir(label_dir):
            if not label_file.endswith(".txt"):
                continue

            base = os.path.splitext(label_file)[0]
            suffix = dataset_folder.replace("/", "_")
            new_base = f"{base}_{suffix}"

            # Chequear longitud del nuevo nombre base
            too_long = False
            for ext in ['.jpg', '.jpeg', '.png', '.txt']:
                if len(new_base + ext) > MAX_FILENAME_LENGTH:
                    too_long = True
            if too_long:
                print(f"⏩ Archivo omitido por nombre largo: {new_base}")
                continue

            old_label_path = os.path.join(label_dir, label_file)
            new_label_path = os.path.join(MERGED_DIR, subset, "labels", new_base + ".txt")

            # Copiar imagen con sufijo también
            copied = False
            for ext in ['.jpg', '.jpeg', '.png']:
                old_img_path = os.path.join(image_dir, base + ext)
                if os.path.exists(old_img_path):
                    new_img_path = os.path.join(MERGED_DIR, subset, "images", new_base + ext)
                    if not is_name_too_long(new_img_path):
                        shutil.copy2(old_img_path, new_img_path)
                        copied = True
                    break

            if not copied:
                print(f"⏩ Imagen no encontrada o nombre largo: {base}")
                continue

            # Copiar y ajustar label
            try:
                with open(old_label_path, 'r') as f_in, open(new_label_path, 'w') as f_out:
                    for line in f_in:
                        parts = line.strip().split()
                        if len(parts) < 5:
                            continue
                        cls = int(parts[0])
                        if dataset_folder == PEOPLE_DIR:
                            new_cls = 0  # person
                        else:
                            new_cls = cls  # top/bottom ya están como 1 o 2
                        f_out.write(f"{new_cls} {' '.join(parts[1:])}\n")
            except Exception as e:
                print(f"⚠️ Error copiando label {label_file}: {e}")
                continue

def create_yaml():
    yaml_path = os.path.join(MERGED_DIR, "data.yaml")
    data = {
        "train": "../train/images",
        "val": "../valid/images",
        "test": "../test/images",
        "nc": len(FINAL_CLASSES),
        "names": FINAL_CLASSES
    }
    with open(yaml_path, 'w') as f:
        yaml.dump(data, f, sort_keys=False)
    print(f"✔️  data.yaml generado en {yaml_path}")

def main():
    print("🛠️  Unificando datasets...")
    ensure_dirs()

    folders = [f for f in os.listdir(".") if f.startswith("Clothes") or f.startswith("clothes") or f.startswith("Clothing") or f == PEOPLE_DIR]

    for folder in folders:
        if os.path.isdir(folder):
            print(f"📂 Procesando: {folder}")
            copy_and_suffix_files(folder)

    create_yaml()
    print("✅ Dataset unificado en:", MERGED_DIR)

if __name__ == "__main__":
    main()
```

## Cuantificando clases

```python
import os
from collections import defaultdict

# Ruta a tu dataset YOLOv8 fusionado
MERGED_DATASET_DIR = "merged-dataset"

# Nombres de las clases en el orden esperado
CLASS_NAMES = ["person", "top", "bottom"]

# Inicializamos contador
detection_counts = defaultdict(int)

# Subcarpetas que recorreremos
SUBSETS = ["train", "valid", "test"]

for subset in SUBSETS:
    label_dir = os.path.join(MERGED_DATASET_DIR, subset, "labels")
    if not os.path.exists(label_dir):
        print(f"⚠️  No se encontró carpeta: {label_dir}")
        continue

    for filename in os.listdir(label_dir):
        if not filename.endswith(".txt"):
            continue

        filepath = os.path.join(label_dir, filename)
        with open(filepath, "r") as f:
            for line in f:
                parts = line.strip().split()
                if not parts or not parts[0].isdigit():
                    continue
                class_id = int(parts[0])
                detection_counts[class_id] += 1

# Mostrar resultados
print("📊 Conteo de detecciones por clase:")
for idx, name in enumerate(CLASS_NAMES):
    count = detection_counts.get(idx, 0)
    print(f"  {idx}: {name:<7} → {count} detecciones")
```

## Data Augmentation

```python
import os
import random
from shutil import move
from collections import defaultdict
from tqdm import tqdm

# 📁 Dataset base
DATASET_DIR = "merged-dataset"
SUBSETS = ["train", "valid", "test"]
TARGET_CLASSES = [1, 2]  # 1: top, 2: bottom
TARGET_TOTAL = 10000
CLASS_NAMES = {1: "top", 2: "bottom"}

# 🧮 Cuántas detecciones por subset (proporcional)
TARGET_SPLIT = {
    "train": int(TARGET_TOTAL * 0.8),
    "valid": int(TARGET_TOTAL * 0.1),
    "test": int(TARGET_TOTAL * 0.1)
}

def get_all_labels(subset):
    """Retorna lista de (label_path, clase, cantidad de detecciones de esa clase)"""
    label_dir = os.path.join(DATASET_DIR, subset, "labels")
    label_info = []

    for fname in os.listdir(label_dir):
        if not fname.endswith(".txt"):
            continue
        full_path = os.path.join(label_dir, fname)
        counts = defaultdict(int)
        with open(full_path) as f:
            for line in f:
                parts = line.strip().split()
                if parts and parts[0].isdigit():
                    cls = int(float(parts[0]))
                    if cls in TARGET_CLASSES:
                        counts[cls] += 1
        for cls, count in counts.items():
            label_info.append((full_path, cls, count))

    return label_info

def reduce_class(subset, cls_id, keep_limit):
    """Reduce una clase en un subset hasta un máximo"""
    print(f"\n🔻 Reduciendo '{CLASS_NAMES[cls_id]}' en '{subset}' a máx {keep_limit} detecciones...")

    labels = get_all_labels(subset)
    random.shuffle(labels)

    # Recolectar etiquetas a mantener
    keep = []
    total = 0
    for path, c, count in labels:
        if c != cls_id:
            continue
        if total + count <= keep_limit:
            keep.append(path)
            total += count
        elif total < keep_limit:
            keep.append(path)
            total = keep_limit
            break
        else:
            break

    print(f"✅ Se mantendrán {total} detecciones en {len(keep)} archivos")

    # Eliminar etiquetas y sus imágenes si no están en 'keep'
    label_dir = os.path.join(DATASET_DIR, subset, "labels")
    image_dir = os.path.join(DATASET_DIR, subset, "images")
    deleted = 0

    for fname in os.listdir(label_dir):
        fpath = os.path.join(label_dir, fname)
        if fpath not in keep:
            # Reescribir solo si contiene esa clase
            new_lines = []
            with open(fpath, 'r') as f:
                for line in f:
                    parts = line.strip().split()
                    if parts and int(float(parts[0])) != cls_id:
                        new_lines.append(line)

            if new_lines:
                with open(fpath, 'w') as f:
                    f.writelines(new_lines)
            else:
                os.remove(fpath)
                # Borrar imagen correspondiente
                base = os.path.splitext(fname)[0]
                for ext in [".jpg", ".jpeg", ".png"]:
                    img_path = os.path.join(image_dir, base + ext)
                    if os.path.exists(img_path):
                        os.remove(img_path)
                        break
                deleted += 1

    print(f"🗑️ Eliminados {deleted} archivos sin la clase {CLASS_NAMES[cls_id]}")

def main():
    print("📉 Iniciando reducción de clases 'top' y 'bottom'...\n")
    for cls in TARGET_CLASSES:
        for subset in SUBSETS:
            reduce_class(subset, cls, TARGET_SPLIT[subset])

    print("\n✅ Reducción completa. Las clases 'top' y 'bottom' ahora tienen ~10k detecciones cada una.")

if __name__ == "__main__":
    main()
```

## Para Top

```python
import os
import cv2
import albumentations as A
from glob import glob
from collections import defaultdict
from tqdm import tqdm

DATASET_DIR = "merged-dataset"
TARGET_CLASS = 1  # top
TOTAL_TARGET = 20000
AUG_LIMIT_PER_FILE = 5
SUBSETS = ["train", "valid", "test"]

transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.4),
    A.Rotate(limit=15, p=0.5),
    A.RandomScale(scale_limit=0.2, p=0.3),
    A.HueSaturationValue(p=0.3)
], bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels']))

subset_counts = {}
subset_targets = {}

def count_current_detections():
    total = 0
    for subset in SUBSETS:
        count = 0
        label_dir = os.path.join(DATASET_DIR, subset, "labels")
        for file in os.listdir(label_dir):
            if file.endswith(".txt"):
                with open(os.path.join(label_dir, file), 'r') as f:
                    for line in f:
                        if line.strip().startswith(f"{TARGET_CLASS} "):
                            count += 1
        subset_counts[subset] = count
        total += count

    # calcular proporciones
    for subset in SUBSETS:
        prop = subset_counts[subset] / max(total, 1)
        subset_targets[subset] = max(0, int(TOTAL_TARGET * prop) - subset_counts[subset])

def load_labels(label_path):
    bboxes = []
    labels = []
    with open(label_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 5 and int(parts[0]) == TARGET_CLASS:
                bboxes.append([float(x) for x in parts[1:]])
                labels.append(TARGET_CLASS)
    return bboxes, labels

def write_yolo_labels(filepath, bboxes, labels):
    with open(filepath, 'w') as f:
        for cls, (x, y, w, h) in zip(labels, bboxes):
            f.write(f"{cls} {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")

def augment_image(img_path, lbl_path, out_img_dir, out_lbl_dir, base_name, subset):
    img = cv2.imread(img_path)
    if img is None:
        return

    bboxes, labels = load_labels(lbl_path)
    if not bboxes:
        return

    for i in range(AUG_LIMIT_PER_FILE):
        if subset_targets[subset] <= 0:
            break

        transformed = transform(image=img, bboxes=bboxes, class_labels=labels)
        new_img = transformed['image']
        new_bboxes = transformed['bboxes']
        new_labels = transformed['class_labels']

        if not new_bboxes:
            continue

        new_name = f"{base_name}_aug{i}"
        img_out = os.path.join(out_img_dir, new_name + ".jpg")
        lbl_out = os.path.join(out_lbl_dir, new_name + ".txt")

        cv2.imwrite(img_out, new_img)
        write_yolo_labels(lbl_out, new_bboxes, new_labels)

        subset_targets[subset] -= len(new_labels)

def main():
    print("📊 Contando detecciones actuales de 'top'...")
    count_current_detections()

    print("\n🎯 Objetivo por subset:")
    for subset in SUBSETS:
        print(f"  {subset:<5}: objetivo adicional → {subset_targets[subset]}")

    for subset in SUBSETS:
        print(f"\n🚀 Augmentando '{subset}'...")
        label_dir = os.path.join(DATASET_DIR, subset, "labels")
        image_dir = os.path.join(DATASET_DIR, subset, "images")
        label_files = glob(os.path.join(label_dir, "*.txt"))

        for lbl_path in tqdm(label_files):
            if subset_targets[subset] <= 0:
                break

            base = os.path.splitext(os.path.basename(lbl_path))[0]
            img_path = None
            for ext in [".jpg", ".jpeg", ".png"]:
                candidate = os.path.join(image_dir, base + ext)
                if os.path.exists(candidate):
                    img_path = candidate
                    break
            if img_path:
                augment_image(img_path, lbl_path, image_dir, label_dir, base, subset)

    print("\n✅ Augmentación finalizada.")
    print("📦 Detecciones totales deseadas de clase 'top':", TOTAL_TARGET)

if __name__ == "__main__":
    main()
```

## Para Bottom

```python
import os
import cv2
import albumentations as A
from glob import glob
from collections import defaultdict
from tqdm import tqdm

DATASET_DIR = "merged-dataset"
TARGET_CLASS = 2  # bottom
TOTAL_TARGET = 19990
AUG_LIMIT_PER_FILE = 5
SUBSETS = ["train", "valid", "test"]

transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.4),
    A.Rotate(limit=15, p=0.5),
    A.RandomScale(scale_limit=0.2, p=0.3),
    A.HueSaturationValue(p=0.3)
], bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels']))

subset_counts = {}
subset_targets = {}

def count_current_detections():
    total = 0
    for subset in SUBSETS:
        count = 0
        label_dir = os.path.join(DATASET_DIR, subset, "labels")
        for file in os.listdir(label_dir):
            if file.endswith(".txt"):
                with open(os.path.join(label_dir, file), 'r') as f:
                    for line in f:
                        if line.strip().startswith(f"{TARGET_CLASS} "):
                            count += 1
        subset_counts[subset] = count
        total += count

    # calcular proporciones
    for subset in SUBSETS:
        prop = subset_counts[subset] / max(total, 1)
        subset_targets[subset] = max(0, int(TOTAL_TARGET * prop) - subset_counts[subset])

def load_labels(label_path):
    bboxes = []
    labels = []
    with open(label_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 5 and int(float(parts[0])) == TARGET_CLASS:
                bboxes.append([float(x) for x in parts[1:]])
                labels.append(TARGET_CLASS)
    return bboxes, labels

def write_yolo_labels(filepath, bboxes, labels):
    with open(filepath, 'w') as f:
        for cls, (x, y, w, h) in zip(labels, bboxes):
            f.write(f"{cls} {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")

def augment_image(img_path, lbl_path, out_img_dir, out_lbl_dir, base_name, subset):
    img = cv2.imread(img_path)
    if img is None:
        return

    bboxes, labels = load_labels(lbl_path)
    if not bboxes:
        return

    for i in range(AUG_LIMIT_PER_FILE):
        if subset_targets[subset] <= 0:
            break

        transformed = transform(image=img, bboxes=bboxes, class_labels=labels)
        new_img = transformed['image']
        new_bboxes = transformed['bboxes']
        new_labels = transformed['class_labels']

        if not new_bboxes:
            continue

        new_name = f"{base_name}_aug_btm{i}"
        img_out = os.path.join(out_img_dir, new_name + ".jpg")
        lbl_out = os.path.join(out_lbl_dir, new_name + ".txt")

        cv2.imwrite(img_out, new_img)
        write_yolo_labels(lbl_out, new_bboxes, new_labels)

        subset_targets[subset] -= len(new_labels)

def main():
    print("📊 Contando detecciones actuales de 'bottom'...")
    count_current_detections()

    print("\n🎯 Objetivo por subset:")
    for subset in SUBSETS:
        print(f"  {subset:<5}: objetivo adicional → {subset_targets[subset]}")

    for subset in SUBSETS:
        print(f"\n🚀 Augmentando '{subset}'...")
        label_dir = os.path.join(DATASET_DIR, subset, "labels")
        image_dir = os.path.join(DATASET_DIR, subset, "images")
        label_files = glob(os.path.join(label_dir, "*.txt"))

        for lbl_path in tqdm(label_files):
            if subset_targets[subset] <= 0:
                break

            base = os.path.splitext(os.path.basename(lbl_path))[0]
            img_path = None
            for ext in [".jpg", ".jpeg", ".png"]:
                candidate = os.path.join(image_dir, base + ext)
                if os.path.exists(candidate):
                    img_path = candidate
                    break
            if img_path:
                augment_image(img_path, lbl_path, image_dir, label_dir, base, subset)

    print("\n✅ Augmentación finalizada.")
    print("📦 Detecciones totales deseadas de clase 'bottom':", TOTAL_TARGET)

if __name__ == "__main__":
    main()
```

## Comprobando Balanceo

```python
import os
from collections import defaultdict

DATASET_DIR = "merged-dataset"
SUBSETS = ["train", "valid", "test"]
CLASS_NAMES = {0: "person", 1: "top", 2: "bottom"}

def contar_detecciones():
    total_counts = defaultdict(int)
    subset_counts = {subset: defaultdict(int) for subset in SUBSETS}

    for subset in SUBSETS:
        label_dir = os.path.join(DATASET_DIR, subset, "labels")
        if not os.path.exists(label_dir):
            continue
        for file in os.listdir(label_dir):
            if file.endswith(".txt"):
                with open(os.path.join(label_dir, file), 'r') as f:
                    for line in f:
                        parts = line.strip().split()
                        if not parts or not parts[0].replace('.', '', 1).isdigit():
                            continue
                        cls = int(float(parts[0]))
                        if cls in CLASS_NAMES:
                            total_counts[cls] += 1
                            subset_counts[subset][cls] += 1

    return total_counts, subset_counts

def mostrar_reporte():
    total, subset = contar_detecciones()

    print("📊 Detecciones por clase:")
    for cls_id in sorted(CLASS_NAMES.keys()):
        print(f"  {cls_id}: {CLASS_NAMES[cls_id]:<7} → {total[cls_id]} detecciones")

    print("\n📁 Detecciones por subset:")
    for s in SUBSETS:
        print(f"  📂 {s}:")
        for cls_id in sorted(CLASS_NAMES.keys()):
            print(f"    - {CLASS_NAMES[cls_id]:<7}: {subset[s][cls_id]}")

if __name__ == "__main__":
    mostrar_reporte()
```

## Congelamiento de capas para utilizar como extractor de caracteristicas

```python
# ✅ Callback personalizado para congelar las primeras 10 capas al iniciar entrenamiento
def freeze_layer(trainer):
    model = trainer.model
    num_freeze = 8
    print(f"\n[INFO] Congelando las primeras {num_freeze} capas...\n")

    freeze_prefixes = [f'model.{i}.' for i in range(num_freeze)]

    for name, param in model.named_parameters():
        param.requires_grad = True  # por defecto entrenable
        if any(prefix in name for prefix in freeze_prefixes):
            param.requires_grad = False
            print(f"❄️ {name} congelado")

    print(f"\n✅ Se congelaron {num_freeze} capas correctamente.\n")

# Registrar el callback
model.add_callback("on_train_start", freeze_layer)
```
